```python
xonar_soni = [1, 2, 3, 4, 5, 6, 7, 8]  # Xonalar soni
narxi = [15000, 25000, 32000, 45000, 60000, 75000, 90000, 110000]  # Narxlar
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Yangi ma'lumotlar
xonar_soni = [1, 2, 3, 4, 5, 6, 7, 8]
narxi = [15000, 25000, 32000, 45000, 60000, 75000, 90000, 110000]

# Ma'lumotlarni numpy array sifatida tayyorlash
X = np.array(xonar_soni).reshape(-1, 1)
y = np.array(narxi)

# Lineer regressiya modelini yaratish va o'qitish
model = LinearRegression()
model.fit(X, y)

# Natijalarni chiqarish
print("Regressiya koeffitsiyenti (beta_1):", model.coef_)
print("Regressiya o'zgaruvchisi (beta_0):", model.intercept_)

# Grafikni chizish
plt.scatter(xonar_soni, narxi, color='blue', label='Haqiqiy ma\'lumotlar')
plt.plot(xonar_soni, model.predict(X), color='red', label='Regressiya chizig\'i')
plt.xlabel('Xonar soni')
plt.ylabel('Narxi')
plt.title('Xonar soni va narxi o\'rtasidagi bog\'liqlik')
plt.legend()
plt.show()
```
